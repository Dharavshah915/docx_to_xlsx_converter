```python
!pip install PyDocX
!pip install PyDocX beautifulsoup4
from bs4 import BeautifulSoup
import re
import pandas as pd
from pydocx import PyDocX
from google.colab import files
import collections.abc

collections.Hashable = collections.abc.Hashable

doc_path = "24B-D.docx"  # Update this path as needed, make sure file is .docx and uploaded
output_path = "24B-D_converted.xlsx" # Change output path as needed



#conversion
html_content = PyDocX.to_html(doc_path)

soup = BeautifulSoup(html_content, "html.parser")
text = soup.get_text(separator="\n")

lines = [line.strip() for line in text.splitlines() if line.strip()]

pattern = re.compile( #regex needs to be changed based on what information needs to be captured
    r'(\d+[A-Z]?\d*)\s+(.+?)\s+(1:\d+(?:,\d+)*)\s+(-|\d+)\s+([A-Z]+\.\s?\d{0,2}/?\d{0,2}/?\d{2,4})'
)

data_rows = []

for line in lines:
    match = pattern.match(line)
    if match:
        chart_number = match.group(1)
        title = match.group(2)
        scale = match.group(3)
        edition = match.group(4)
        date = match.group(5)
        data_rows.append([chart_number, title, scale, edition, date])

columns = ["Chart Number", "Sheet Title", "Scale", "Edition", "Date"]
df = pd.DataFrame(data_rows, columns=columns)

df.to_excel(output_path, index=False)
files.download(output_path)

print(f"Data successfully extracted and saved to: {output_path}")
```
